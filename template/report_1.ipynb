## 2. NDAA Tof Sensor


Two potential sensors were found, both built by Broadcom that could potentially meet the requirements for Jerry's ToF sensor. I reached out and they noted the ArkElectronics breakout boards as examples of NDAA uses

- **[AFBR-S50LV85D](https://www.broadcom.com/products/optical-sensors/time-of-flight-3d-sensors/afbr-s50lv85d):** 30m range standard, up to 100m in dual-frequency mode (capped at 100 Hz frame rate at long range; up to 3 kHz in short-range single-frequency mode). 12.4° × 6.2° FoV, 32 pixels, SPI. ~$60/unit single-qty on DigiKey.
- **[AFBR-S50LX85D](https://www.broadcom.com/products/optical-sensors/time-of-flight-3d-sensors/afbr-s50lx85d):** 50mm to 50m range, same 100 Hz cap in dual-frequency long-range mode, up to 3 kHz short-range. Optimized for drone and industrial long-range sensing. ~$87–91/unit single-qty on DigiKey.
